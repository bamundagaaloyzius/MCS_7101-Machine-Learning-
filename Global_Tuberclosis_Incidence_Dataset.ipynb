- **DIM_GEO_CODE_M49** : The geographical code
- **DIM_GEO_CODE_TYPE** : The geographical entity
- **DIM_PUBLISH_STATE_CODE** : This states whether the data is published by WHO or NOt.
- **IND_NAME**: Disease case for this case Tuberclosis
- **GEO_NAME_SHORT** : country or region of data collection
- **RATE_PER_100000_N** : Average number of incidence cases for every 100,000 people for that year.
- **RATE_PER_100000_NL** : Maximum number of incidence cases for every 100,000 people for that year.
- **RATE_PER_100000_NU** : Minimum number of incidence cases for every 100,000 people for that year.

```python
from sklearn.model_selection import train_test_split

X = data.drop(columns=['Country','TB Incidence'])
y = data['TB Incidence']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=8)
```

```python
from sklearn.pipeline import Pipeline
from category_encoders.target_encoder import TargetEncoder
from xgboost import XGBRegressor

estimators = [
    ('encoder', TargetEncoder()),
    ('reg', XGBRegressor(random_state=8)) # can customize objective function with the objective parameter
]
pipe = Pipeline(steps=estimators)
pipe
```

```python
from skopt import BayesSearchCV
from skopt.space import Real, Categorical, Integer

search_space = {
    'reg__max_depth': Integer(2,8),
    'reg__learning_rate': Real(0.001, 1.0, prior='log-uniform'),
    'reg__subsample': Real(0.5, 1.0),
    'reg__colsample_bytree': Real(0.5, 1.0),
    'reg__colsample_bylevel': Real(0.5, 1.0),
    'reg__colsample_bynode' : Real(0.5, 1.0),
    'reg__reg_alpha': Real(0.0, 10.0),
    'reg__reg_lambda': Real(0.0, 10.0),
    'reg__gamma': Real(0.0, 10.0)
}

# Change scoring to a regression metric like 'neg_mean_squared_error' or 'r2'
opt = BayesSearchCV(pipe, search_space, cv=3, n_iter=10, scoring='neg_mean_squared_error', random_state=8)
# in reality, you may consider setting cv and n_iter to higher values
```

```python
opt.fit(X_train, y_train)
```

```python
opt.best_estimator_
```

```python
opt.best_score_
```

```python
opt.score(X_test, y_test)
```

```python
opt.predict(X_test)
```

```python
opt.best_estimator_.steps
```

```python
from xgboost import plot_importance

xgboost_step = opt.best_estimator_.steps[1]
xgboost_model = xgboost_step[1]
plot_importance(xgboost_model)
```
